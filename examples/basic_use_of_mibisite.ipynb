## Initialize Mibisite object

```python
from mibisite import Mibisite

mysite = Mibisite("eScienceCenter", "perimeter.shp", "landmarks.shp")
mysite.load_wells("wells.csv")

mysite.show_map()
```

```python
import mibiscreen as mbs

mysite.load_samples("samples.xlsx")
mbs.activity_plot(mysite)
```

```python
import mibitrans as mbt

mysite.add_source({
    "contaminants":"BTEXIN",
    "coordinates":"4.95686988566346,52.356559093666",
    "depth":"2m",
    "mass":"3000kg",
})

mymodel = mbt.Mibitrans(mysite)
mymodel_results = mymodel.run(simulation_time=10*365)
mysite.add_model_results(mymodel_results)

mysite.show_map()
```

```python
mysite.save_sikb("eScienceCenter_sikb0101.xml")
```
